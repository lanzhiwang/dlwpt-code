## n.Module

```python
class Net(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(6, 8, kernel_size=2)
        self.act1 = nn.Tanh()
        self.fc1 = nn.Linear(72, 32)
        self.act2 = nn.Tanh()
        self.fc2 = nn.Linear(32, 2)

    def forward(self, x):
        out = self.conv1(x)
        out = self.act1(out)  # torch.Size([2, 8, 3, 3])
        out = out.view(out.shape[0], -1)
        out = self.fc1(out)
        out = self.act2(out)
        out = self.fc2(out)

        return out
```

```python
model = Net()
print("model:", model)
print("---" * 3)

# for p in model.parameters():
for p in model.named_parameters():
    print("p:", p)

print("---" * 3)

# print("model.weight:", model.weight)
# print("model.weight.shape:", model.weight.shape)
# print("model.weight.numel:", model.weight.numel())
# print("model.weight.dtype:", model.weight.dtype)
# print("model.weight.device:", model.weight.device)

# print("model.bias:", model.bias)
# print("model.bias.shape:", model.bias.shape)
# print("model.bias.numel:", model.bias.numel())
# print("model.bias.dtype:", model.bias.dtype)
# print("model.bias.device:", model.bias.device)
```

```python
input = torch.randn(2, 6, 4, 4)
# print("input:", input)
print("input.shape:", input.shape)
```
