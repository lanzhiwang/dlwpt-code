## 3.4 命名张量

```python
import torch
```

```python
img_t = torch.randn(3, 5, 5)  # shape [channels, rows, columns]
weights = torch.tensor([0.2126, 0.7152, 0.0722])  # torch.Size([3])
```

```python
batch_t = torch.randn(2, 3, 5, 5)  # shape [batch, channels, rows, columns]
```

```python
# mean 函数的使用方法参考 reference.md
img_gray_naive = img_t.mean(-3)
batch_gray_naive = batch_t.mean(-3)
img_gray_naive.shape, batch_gray_naive.shape
```

```python
unsqueezed_weights = weights.unsqueeze(-1).unsqueeze_(-1)  # torch.Size([3, 1, 1])
img_weights = (
    img_t * unsqueezed_weights
)  # torch.Size([3, 5, 5]) * torch.Size([3, 1, 1])
batch_weights = (
    batch_t * unsqueezed_weights
)  # torch.Size([2, 3, 5, 5]) * torch.Size([3, 1, 1])
img_gray_weighted = img_weights.sum(-3)  # torch.Size([5, 5])
batch_gray_weighted = batch_weights.sum(-3)  # torch.Size([2, 5, 5])
batch_weights.shape, batch_t.shape, unsqueezed_weights.shape
```

```python
"""
>>> img_t = torch.randn(3, 5, 5)
>>> img_t.size()
torch.Size([3, 5, 5])
>>> weights = torch.tensor([0.2126, 0.7152, 0.0722])
>>> weights.size()
torch.Size([3])
>>> batch_t = torch.randn(2, 3, 5, 5)
>>> batch_t.size()
torch.Size([2, 3, 5, 5])
>>> img_gray_weighted_fancy = torch.einsum("...chw,c->...hw", img_t, weights)
>>> img_gray_weighted_fancy.size()
torch.Size([5, 5])
>>> batch_gray_weighted_fancy = torch.einsum("...chw,c->...hw", batch_t, weights)
>>> batch_gray_weighted_fancy.size()
torch.Size([2, 5, 5])
>>>
"""

img_gray_weighted_fancy = torch.einsum("...chw,c->...hw", img_t, weights)
batch_gray_weighted_fancy = torch.einsum("...chw,c->...hw", batch_t, weights)
batch_gray_weighted_fancy.shape
```

```python
weights_named = torch.tensor([0.2126, 0.7152, 0.0722], names=["channels"])
weights_named
```

```python
img_named = img_t.refine_names(..., "channels", "rows", "columns")
batch_named = batch_t.refine_names(..., "channels", "rows", "columns")
print("img named:", img_named.shape, img_named.names)
print("batch named:", batch_named.shape, batch_named.names)
```

```python
# align_as() 方法返回一个张量, 其中添加了缺失的维度
weights_aligned = weights_named.align_as(img_named)
weights_aligned.shape, weights_aligned.names
```

```python
gray_named = (img_named * weights_aligned).sum("channels")
gray_named.shape, gray_named.names
```

```python
try:
    gray_named = (img_named[..., :3] * weights_named).sum("channels")
except Exception as e:
    print(e)
```

```python
gray_plain = gray_named.rename(None)
gray_plain.shape, gray_plain.names
```
